```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score
import skfuzzy as fuzz
# Load iris dataset
iris = load_iris()
X = iris.data.T   # skfuzzy expects features as rows and samples as columns
# ------------------------------
# Run Fuzzy C-Means
# ------------------------------
n_clusters = 3
cntr, u, u0, d, jm, p, fpc = fuzz.cluster.cmeans(
    X, c=n_clusters, m=2, error=0.005, maxiter=1000, init=None, seed=42
)
# Membership matrix (u) -> shape (n_clusters, n_samples)
labels = np.argmax(u, axis=0)  # Hard assignment by max membership
# ------------------------------
# Evaluation
# ------------------------------
sil_score = silhouette_score(X.T, labels)
print(f"Fuzzy C-Means: Silhouette Score = {sil_score:.3f}")
print(f"Partition Coefficient (FPC) = {fpc:.3f}")
# ------------------------------
# PCA Visualization
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X.T)
# Cluster centers in PCA space
centers_pca = pca.transform(cntr)
plt.figure(figsize=(8,6))
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=labels, cmap='coolwarm', s=50)
plt.scatter(centers_pca[:, 0], centers_pca[:, 1],
            marker='X', s=200, c='black', label='Centers')
plt.title("Fuzzy C-Means Clustering on Iris (PCA 2D)")
plt.xlabel("PCA Component 1")
plt.ylabel("PCA Component 2")
plt.legend()
plt.show()
'''Output -
Fuzzy C-Means: Silhouette Score = 0.550
Partition Coefficient (FPC) = 0.783
'''
```
