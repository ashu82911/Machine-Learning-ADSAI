```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.cluster import DBSCAN
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score
# Load iris dataset (only features, no labels)
iris = load_iris()
X = iris.data
# ------------------------------
# Run DBSCAN
# ------------------------------
dbscan = DBSCAN(eps=0.8, min_samples=5)  # tweak eps to change clusters
labels = dbscan.fit_predict(X)
# Number of clusters (excluding noise)
n_clusters = len(set(labels)) - (1 if -1 in labels else 0)
print(f"DBSCAN found {n_clusters} clusters (excluding noise).")
# Silhouette Score (only if >= 2 clusters)
if n_clusters > 1:
    sil_score = silhouette_score(X, labels)
    print(f"Silhouette Score: {sil_score:.3f}")
else:
    print("Silhouette Score not defined (only 1 cluster found).")
# ------------------------------
# PCA visualization of DBSCAN clusters
# ------------------------------
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)
plt.figure(figsize=(8,6))
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=labels, cmap='plasma', s=50)
plt.title(f"DBSCAN Clustering on Iris (PCA 2D)\nClusters found: {n_clusters}")
plt.xlabel("PCA Component 1")
plt.ylabel("PCA Component 2")
plt.colorbar(label="Cluster Label")
plt.show()
'''Output -
DBSCAN found 2 clusters (excluding noise).
Silhouette Score: 0.512
'''
```
