```python
import numpy as np
import matplotlib.pyplot as plt
import skfuzzy as fuzz
import imageio
import os
# Create synthetic dataset
np.random.seed(42)
n_samples = 150
X = np.vstack([
    np.random.normal([2, 2], 0.3, (n_samples, 2)),
    np.random.normal([6, 2], 0.3, (n_samples, 2)),
    np.random.normal([4, 6], 0.3, (n_samples, 2))
]).T  # Shape: (2, total_samples)
# Parameters
n_clusters = 3
m = 2
error = 0.005
maxiter = 30  # We'll stop early to visualize
# Initialize cluster centers randomly
cntr = np.random.rand(n_clusters, X.shape[0])
frames = []
# Custom iterative process to save frames
for iteration in range(1, maxiter + 1):
    # Run clustering for a set number of iterations
    cntr, u, _, _, _, _, _ = fuzz.cluster.cmeans(
        X, c=n_clusters, m=m, error=error, maxiter=iteration, init=None
    )
    hard_labels = np.argmax(u, axis=0)
    # Plot
    fig, ax = plt.subplots()
    colors = ['red', 'blue', 'green']
    for i in range(n_clusters):
        ax.scatter(X[0, hard_labels == i], X[1, hard_labels == i],
                   c=colors[i], label=f'Cluster {i+1}', alpha=0.6)
    ax.scatter(cntr[:, 0], cntr[:, 1], c='black', marker='X', s=200, label='Centers')
    ax.set_title(f'Fuzzy C-Means Iteration {iteration}')
    ax.legend()
    # Save frame
    fname = f'fcm_frame_{iteration}.png'
    plt.savefig(fname)
    plt.close()
    frames.append(imageio.imread(fname))
# Save GIF
imageio.mimsave('fcm_clustering.gif', frames, duration=0.5)
# Cleanup temp files
for iteration in range(1, maxiter + 1):
    os.remove(f'fcm_frame_{iteration}.png')
print("GIF saved as fcm_clustering.gif")
```
