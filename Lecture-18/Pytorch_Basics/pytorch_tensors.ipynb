```python
import torch
# Creating a 1D tensor
x = torch.tensor([1.0, 2.0, 3.0])
print('1D Tensor: \n', x)
# Creating a 2D tensor
y = torch.zeros((3, 3))
print('2D Tensor: \n', y)
'''Output -
1D Tensor:
tensor([1., 2., 3.])
2D Tensor:
tensor([[0., 0., 0.],
       [0., 0., 0.],
       [0., 0., 0.]])
'''
```
