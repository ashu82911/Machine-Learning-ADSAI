```python
import numpy as np
from sklearn.neighbors import KNeighborsRegressor
# Original dataset
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2.2, 3.9, 6.1, 7.8, 10.2])
# Point to predict
x_test = np.array([[3]])
# --- Single KNN Regressor ---
knn = KNeighborsRegressor(n_neighbors=3)
knn.fit(X, y)
single_knn_pred = knn.predict(x_test)[0]
print(f"Single KNN prediction at x=3: {single_knn_pred:.2f}")
# --- Manually Defined Bootstrap Samples ---
# Each sample is a list of indices (with replacement)
bootstrap_indices_list = [
   [0, 1, 2, 3, 3],  # Sample 1: [1, 2, 3, 4, 4]
   [1, 2, 2, 4, 0],  # Sample 2: [2, 3, 3, 5, 1]
   [2, 3, 4, 4, 1]   # Sample 3: [3, 4, 5, 5, 2]
]
bagging_preds = []
# Fit 3 KNN regressors on bootstrap samples
for i, indices in enumerate(bootstrap_indices_list):
   X_sample = X[indices]
   y_sample = y[indices]
   knn_model = KNeighborsRegressor(n_neighbors=3)
   knn_model.fit(X_sample, y_sample)
   pred = knn_model.predict(x_test)[0]
   bagging_preds.append(pred)
   # Show model-wise prediction
   print(f"Model {i+1} prediction at x=3 (bootstrap indices {indices}): {pred:.2f}")
# Average of all bagging predictions
bagging_final_pred = np.mean(bagging_preds)
print(f"\nFinal Bagging KNN prediction at x=3: {bagging_final_pred:.2f}")
'''Output -
Single KNN prediction at x=3: 5.93
Model 1 prediction at x=3 (bootstrap indices [0, 1, 2, 3, 3]): 5.93
Model 2 prediction at x=3 (bootstrap indices [1, 2, 2, 4, 0]): 5.37
Model 3 prediction at x=3 (bootstrap indices [2, 3, 4, 4, 1]): 5.93
Final Bagging KNN prediction at x=3: 5.74
'''
```
