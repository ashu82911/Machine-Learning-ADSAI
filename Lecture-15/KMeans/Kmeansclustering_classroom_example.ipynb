```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
# Step 1: Define dataset
data = np.array([
   [1, 2],  # A
   [1, 4],  # B
   [1, 0],  # C
   [10, 2], # D
   [10, 4], # E
   [10, 0]  # F
])
# Step 2: Create and fit KMeans model
kmeans = KMeans(n_clusters=3, random_state=0)
kmeans.fit(data)
# Step 3: Get cluster labels and centroids
labels = kmeans.labels_
centroids = kmeans.cluster_centers_
print("Cluster Centers:")
print(centroids)
print("\nCluster Assignments:")
for i, label in enumerate(labels):
   print(f"Point {data[i]} => Cluster {label}")
# Step 4: Predict a new point
new_point = np.array([[2, 3]])
predicted_cluster = kmeans.predict(new_point)
print(f"\nNew point {new_point[0]} assigned to Cluster {predicted_cluster[0]}")
# Step 5: Visualization
plt.figure(figsize=(8, 6))
# Plot clustered points
plt.scatter(data[:, 0], data[:, 1], c=labels, cmap='viridis', label='Data Points')
# Plot centroids
plt.scatter(centroids[:, 0], centroids[:, 1], c='red', marker='X', s=200, edgecolor='black', label='Centroids')
# Plot new point
plt.scatter(new_point[:, 0], new_point[:, 1], c='green', marker='o', s=150, label='New Point')
plt.title('K-Means Clustering with scikit-learn (k=2)')
plt.legend()
plt.grid(True)
plt.show()
'''Output -
Cluster Centers:
[[10.  2.]
[ 1.  2.]]
Cluster Assignments:
Point [1 2] => Cluster 1
Point [1 4] => Cluster 1
Point [1 0] => Cluster 1
Point [10  2] => Cluster 0
Point [10  4] => Cluster 0
Point [10  0] => Cluster 0
New point [2 3] assigned to Cluster 1
'''

```
