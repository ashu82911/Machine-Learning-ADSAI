```python
import numpy as np
from fcmeans import FCM
# Training data
X = np.array([[1], [2], [5]])
# Train FCM
fcm = FCM(n_clusters=2, m=2, max_iter=150)
fcm.fit(X)
# Results
print("Centroids:", fcm.centers)
print("Membership:\n", fcm.u)
# Test new point
new_point = np.array([[3]])
u_new = fcm.predict(new_point)
print("Membership for x=3:", u_new)
'''Output -
Centroids: [[4.99678714]
 [1.49348226]]
Membership:
 [[1.50158630e-02 9.84984137e-01]
 [2.77743931e-02 9.72225607e-01]
 [9.99999160e-01 8.39522826e-07]]
Membership for x=3: [1]
'''
```
