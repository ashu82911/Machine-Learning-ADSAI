```python
import numpy as np
import skfuzzy as fuzz
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap
# Generate synthetic data
np.random.seed(0)
data = np.random.rand(2, 100)
n_clusters = 3
# Perform Fuzzy C-Means clustering
cntr, u, _, _, _, _, fpc = fuzz.cluster.cmeans(
    data, c=n_clusters, m=2, error=0.005, maxiter=1000, init=None
)
hard_clusters = np.argmax(u, axis=0)
print("Cluster Centers:\n", cntr)
print("\nFuzzy Membership Matrix (first 5 data points):\n", u[:, :5])
# Custom bright color map (Red, Green, Blue)
colors = ListedColormap(["red", "green", "blue"])
fig, ax = plt.subplots(figsize=(8, 6))
for i in range(n_clusters):
    sc = ax.scatter(data[0], data[1], c=u[i], cmap=colors,
                    alpha=0.6, label=f'Cluster {i+1}')
# Plot cluster centers
ax.scatter(cntr[:, 0], cntr[:, 1], c='black', marker='X', s=200, label='Cluster Centers')
ax.set_title('Fuzzy C-Means Clustering')
ax.set_xlabel('Feature 1')
ax.set_ylabel('Feature 2')
ax.legend()
plt.show()
```
