```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
import skfuzzy as fuzz
from sklearn.metrics import silhouette_score
# Step 1: Generate an unsupervised dataset (features only)
X, _ = make_blobs(n_samples=300, n_features=2, centers=3, cluster_std=1.2, random_state=42)
# Step 2: Transpose data for scikit-fuzzy (needs shape: [features, samples])
X_T = X.T
# Step 3: Number of clusters
n_clusters = 3
# Step 4: Run Fuzzy C-Means
cntr, u, u0, d, jm, p, fpc = fuzz.cluster.cmeans(
    X_T, c=n_clusters, m=2, error=0.005, maxiter=1000, init=None
)
# Step 5: Convert fuzzy membership to hard labels for evaluation
labels = np.argmax(u, axis=0)
# Step 6: Silhouette score (optional, just to check separation)
sil_score = silhouette_score(X, labels)
print(f"Silhouette Score (hard clusters): {sil_score:.3f}")
print("Cluster Centers:\n", cntr)
# Step 7: Visualization
plt.figure(figsize=(8, 6))
for i in range(n_clusters):
    plt.scatter(X[labels == i, 0], X[labels == i, 1], alpha=0.6, label=f'Cluster {i+1}')
plt.scatter(cntr[:, 0], cntr[:, 1], c='black', marker='X', s=200, label='Centers')
plt.title('Fuzzy C-Means Clustering on Synthetic Dataset')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.show()
```
