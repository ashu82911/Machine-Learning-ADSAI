```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import DBSCAN
# Step 1: Create simple dataset
X_train = np.array([
   [1, 2],  # A
   [2, 2],  # B
   [2, 3],  # C
   [8, 7],  # D
   [8, 8],  # E
   [25, 80] # F (Noise)
])
# New test point
test_point = np.array([[2, 1.5]])
# Combine for prediction
X_full = np.vstack([X_train, test_point])
# Step 2: Apply DBSCAN
dbscan = DBSCAN(eps=1.5, min_samples=2)
labels = dbscan.fit_predict(X_full)
# Step 3: Plot clusters
plt.figure(figsize=(8, 6))
# Colors for clusters
colors = ['blue', 'green', 'red', 'purple', 'orange']
unique_labels = set(labels)
for label in unique_labels:
   cluster_points = X_full[labels == label]
   if label == -1:
       plt.scatter(cluster_points[:, 0], cluster_points[:, 1],
       c='black', label='Noise', s=100, marker='x')
   else:
       plt.scatter(cluster_points[:, 0], cluster_points[:, 1],
       c=colors[label % len(colors)], label=f'Cluster {label}', s=100)
# Highlight the test point
plt.scatter(test_point[0][0], test_point[0][1],c='yellow', edgecolors='black', s=200, label='Test Point')
plt.title('DBSCAN on Simple 2D Dataset')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid(True)
plt.show()
```
