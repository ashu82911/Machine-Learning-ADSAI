```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.preprocessing import StandardScaler
# Sample dataset
data = {
   'Age':    [25, 45, 35, 22, 42, 28],
   'Income': [40, 80, 60, 35, 70, 50],
   'Buy':    [0, 1, 1, 0, 1, 0]
}
df = pd.DataFrame(data)
# Features and target
X = df[['Age', 'Income']]
y = df['Buy']
# Standardize features (important for gradient boosting sometimes)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
# Model: Stochastic Gradient Boosting
model = GradientBoostingClassifier(
   learning_rate=0.1,
   n_estimators=20,
   subsample=0.5,  # stochastic!
   max_depth=1,
   random_state=0
)
model.fit(X_scaled, y)
# Evaluate on training data (since we have small dataset)
y_pred = model.predict(X_scaled)
print("Confusion Matrix:\n", confusion_matrix(y, y_pred))
print("\nClassification Report:\n", classification_report(y, y_pred))
# Predict on a new sample
# New data point: Age=30, Income=55
new_data = np.array([[30, 55]])
new_data_scaled = scaler.transform(new_data)
pred = model.predict(new_data_scaled)
proba = model.predict_proba(new_data_scaled)
print(f"\nPrediction for Age=30, Income=55: {'Buy' if pred[0] == 1 else 'Not Buy'}")
print(f"Probability of Buy: {proba[0][1]*100:.2f}%")
# Optional: Plot decision boundary
from matplotlib.colors import ListedColormap
def plot_decision_boundary(model, X, y, title):
   X = X.values if isinstance(X, pd.DataFrame) else X
   h = .02
   x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
   y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
   xx, yy = np.meshgrid(np.arange(x_min, x_max, h),
                        np.arange(y_min, y_max, h))
   Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
   Z = Z.reshape(xx.shape)
   cmap_light = ListedColormap(['#FFAAAA', '#AAFFAA'])
   cmap_bold = ['#FF0000', '#00FF00']
   plt.figure(figsize=(8, 6))
   plt.contourf(xx, yy, Z, cmap=cmap_light, alpha=0.5)
   plt.scatter(X[:, 0], X[:, 1], c=y, cmap=ListedColormap(cmap_bold), edgecolor='k', s=60)
   plt.xlabel("Age (standardized)")
   plt.ylabel("Income (standardized)")
   plt.title(title)
   plt.grid(True)
   plt.tight_layout()
   plt.show()
plot_decision_boundary(model, X_scaled, y, "Stochastic Gradient Boosting Decision Boundary")
'''Output -
Confusion Matrix:
[[3 0]
[0 3]]
Classification Report:
              precision    recall  f1-score   support
          0       1.00      1.00      1.00         3
          1       1.00      1.00      1.00         3
   accuracy                           1.00         6
  macro avg       1.00      1.00      1.00         6
weighted avg       1.00      1.00      1.00         6
Prediction for Age=30, Income=55: Not Buy
Probability of Buy: 33.89%
'''
```
