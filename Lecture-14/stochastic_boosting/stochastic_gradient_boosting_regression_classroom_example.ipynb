```python
import numpy as np
import pandas as pd
# Dataset
data = pd.DataFrame({
   'Age':    [25, 30, 35, 40, 45, 50],
   'Income': [40, 60, 50, 80, 100, 90],
   'Price':  [200, 250, 240, 300, 360, 330]
})
X = data[['Age', 'Income']].values
y = data['Price'].values
# Initial prediction: mean
F0 = np.mean(y)
preds = np.full_like(y, fill_value=F0, dtype=np.float64)
print(f"Initial prediction (F0): {F0}")
# Residuals after F0
r1 = y - preds
data['r1'] = r1
# First stump: on Income (Income <= 60)
h1 = np.where(data['Income'] <= 60, -60, 80)
F1 = preds + 0.1 * h1
data['F1'] = F1
# Residuals after F1
r2 = y - F1
data['r2'] = r2
# Second stump: on Age (Age <= 40)
h2 = np.where(data['Age'] <= 40, -6, 42)
F2 = F1 + 0.1 * h2
data['F2'] = F2
# Final residuals
data['FinalResidual'] = y - F2
print("\nFinal table after 2 boosting rounds:")
print(data[['Age', 'Income', 'Price', 'F2', 'FinalResidual']])
# Prediction on new data
new_point = {'Age': 32, 'Income': 55}
# Apply h1
h1_new = -60 if new_point['Income'] <= 60 else 80
F1_new = F0 + 0.1 * h1_new
# Apply h2
h2_new = -6 if new_point['Age'] <= 40 else 42
F2_new = F1_new + 0.1 * h2_new
print(f"\nPrediction for new data point (Age=32, Income=55): {F2_new:.2f}")
'''Output -
Initial prediction (F0): 280.0
Final table after 2 boosting rounds:
  Age  Income  Price     F2  FinalResidual
0   25      40    200  273.4          -73.4
1   30      60    250  273.4          -23.4
2   35      50    240  273.4          -33.4
3   40      80    300  287.4           12.6
4   45     100    360  292.2           67.8
5   50      90    330  292.2           37.8
Prediction for new data point (Age=32, Income=55): 273.40
'''
```
