```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, accuracy_score
# 1. Create synthetic dataset
np.random.seed(42)
n = 100
age = np.random.randint(18, 60, size=n)
income = np.random.randint(10, 100, size=n)
# Rule for buying: Older age + higher income -> more likely to buy
buy_product = ((age > 30) & (income > 50)).astype(int)
# Add some randomness
buy_product[np.random.choice(n, 10, replace=False)] = 1 - buy_product[np.random.choice(n, 10, replace=False)]
# Convert to DataFrame
df = pd.DataFrame({
    'Age': age,
    'Income': income,
    'Buy_Product': buy_product
})
# 2. Split into train-test
X = df[['Age', 'Income']]
y = df['Buy_Product']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)
# 3. Initialize Decision Tree with hyperparameters
clf = DecisionTreeClassifier(
    criterion='entropy',
    max_depth=4,
    min_samples_split=10,
    min_samples_leaf=5,
    max_features=None,
    random_state=1,
    class_weight='balanced'
)
# 4. Train model
clf.fit(X_train, y_train)
# 5. Predict and evaluate
y_pred = clf.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
# 6. Visualize decision tree
plt.figure(figsize=(12, 6))
plot_tree(clf, feature_names=['Age', 'Income'], class_names=['No', 'Yes'], filled=True)
plt.title("Decision Tree Classifier")
plt.show()
'''Output -
Accuracy: 0.95
              precision    recall  f1-score   support
           0       0.91      1.00      0.95        10
           1       1.00      0.90      0.95        10
    accuracy                           0.95        20
   macro avg       0.95      0.95      0.95        20
weighted avg       0.95      0.95      0.95        20
'''
```
