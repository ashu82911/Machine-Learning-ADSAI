```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
# Create a synthetic dataset
# Features: Age, Monthly Hours of Study
# Target: Final Exam Score
np.random.seed(42)
n_samples = 100
age = np.random.randint(18, 30, size=n_samples)
hours_per_month = np.random.normal(40, 10, size=n_samples)
exam_score = 50 + 2 * age + 1.5 * hours_per_month + np.random.normal(0, 10, size=n_samples)
df = pd.DataFrame({
    "Age": age,
    "Monthly Study Hours": hours_per_month,
    "Exam Score": exam_score
})
X = df[["Age", "Monthly Study Hours"]]
y = df["Exam Score"]
# Split the dataset
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
# Train Decision Tree Regressor with hyperparameters
regressor = DecisionTreeRegressor(
    criterion='squared_error',
    max_depth=5,
    min_samples_split=5,
    min_samples_leaf=3,
    max_features=None,
    random_state=42
)
regressor.fit(X_train, y_train)
# Predictions
y_pred = regressor.predict(X_test)
# Evaluation
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
print(f"MSE: {mse}")
print(f"R2 Score: {r2}")
# Plot predictions vs actual
plt.figure(figsize=(8, 5))
plt.scatter(y_test, y_pred, alpha=0.7, color='teal')
plt.plot([y.min(), y.max()], [y.min(), y.max()], 'r--', lw=2)
plt.xlabel("Actual Exam Score")
plt.ylabel("Predicted Exam Score")
plt.title("Decision Tree Regressor: Actual vs Predicted")
plt.grid(True)
plt.tight_layout()
plt.show()
'''Outut -
MSE: 228.9800852833821
R2 Score: 0.37221674121643056
'''
```
