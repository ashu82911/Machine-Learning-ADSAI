```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor, plot_tree
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import mean_squared_error, r2_score
# 1. Create synthetic dataset
np.random.seed(42)
n_samples = 100
age = np.random.randint(18, 30, size=n_samples)
hours_per_month = np.random.normal(40, 10, size=n_samples)
exam_score = 50 + 2 * age + 1.5 * hours_per_month + np.random.normal(0, 10, size=n_samples)
df = pd.DataFrame({
    "Age": age,
    "Monthly Study Hours": hours_per_month,
    "Exam Score": exam_score
})
X = df[["Age", "Monthly Study Hours"]]
y = df["Exam Score"]
# 2. Split dataset
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
# 3. Define hyperparameter grid
param_grid = {
    'criterion': ['squared_error', 'friedman_mse'],
    'max_depth': [3, 4, 5, 6, None],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 3, 5],
    'max_features': [None, 'auto', 'sqrt']
}
# 4. Run GridSearchCV
regressor = DecisionTreeRegressor(random_state=42)
grid_search = GridSearchCV(regressor, param_grid, cv=5, scoring='r2', n_jobs=-1)
grid_search.fit(X_train, y_train)
# 5. Use best model for prediction
best_reg = grid_search.best_estimator_
y_pred = best_reg.predict(X_test)
# 6. Evaluation
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
print("Best Parameters:", grid_search.best_params_)
print(f"MSE: {mse:.2f}")
print(f"R2 Score: {r2:.2f}")
# 7. Plot predictions vs actual
plt.figure(figsize=(8, 5))
plt.scatter(y_test, y_pred, alpha=0.7, color='teal')
plt.plot([y.min(), y.max()], [y.min(), y.max()], 'r--', lw=2)
plt.xlabel("Actual Exam Score")
plt.ylabel("Predicted Exam Score")
plt.title("Decision Tree Regressor (GridSearch): Actual vs Predicted")
plt.grid(True)
plt.tight_layout()
plt.show()
'''Output -
Best Parameters: {'criterion': 'squared_error', 'max_depth': 5, 'max_features': None, 'min_samples_leaf': 5, 'min_samples_split': 2}
MSE: 258.29
R2 Score: 0.29
'''
```
