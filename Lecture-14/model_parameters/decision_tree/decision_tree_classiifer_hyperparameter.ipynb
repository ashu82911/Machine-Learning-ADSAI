```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import classification_report, accuracy_score
# 1. Create synthetic dataset
np.random.seed(42)
n = 100
age = np.random.randint(18, 60, size=n)
income = np.random.randint(10, 100, size=n)
# Rule: Older age + higher income -> more likely to buy
buy_product = ((age > 30) & (income > 50)).astype(int)
# Add randomness
random_indices = np.random.choice(n, 10, replace=False)
buy_product[random_indices] = 1 - buy_product[random_indices]
# Create DataFrame
df = pd.DataFrame({
    'Age': age,
    'Income': income,
    'Buy_Product': buy_product
})
# 2. Train-test split
X = df[['Age', 'Income']]
y = df['Buy_Product']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)
# 3. GridSearchCV for hyperparameter tuning
param_grid = {
    'criterion': ['gini', 'entropy'],
    'max_depth': [2, 3, 4, 5],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 3, 5],
    'class_weight': [None, 'balanced']
}
clf = DecisionTreeClassifier(random_state=1)
grid_search = GridSearchCV(clf, param_grid, cv=5, scoring='accuracy')
grid_search.fit(X_train, y_train)
print("Best Parameters:", grid_search.best_params_)
# 4. Evaluate best model
best_clf = grid_search.best_estimator_
y_pred = best_clf.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
# 5. Visualize best decision tree
plt.figure(figsize=(12, 6))
plot_tree(best_clf, feature_names=['Age', 'Income'], class_names=['No', 'Yes'], filled=True)
plt.title("Optimized Decision Tree Classifier (GridSearch)")
plt.show()
'''Output -
Best Parameters: {'class_weight': 'balanced', 'criterion': 'gini', 'max_depth': 2, 'min_samples_leaf': 5, 'min_samples_split': 2}
Accuracy: 0.9
              precision    recall  f1-score   support
           0       1.00      0.82      0.90        11
           1       0.82      1.00      0.90         9
    accuracy                           0.90        20
   macro avg       0.91      0.91      0.90        20
weighted avg       0.92      0.90      0.90        20
'''
```
