```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import mean_squared_error, r2_score
# 1. Generate better synthetic dataset
np.random.seed(42)
n_samples = 200
years_experience = np.random.uniform(0, 15, size=n_samples)  # 0 to 15 years
num_certifications = np.random.randint(0, 10, size=n_samples)  # 0 to 9 certs
# True salary model (add noise)
# Assume: Each year adds ~₹30k, each cert adds ~₹10k, base salary = ₹20k
monthly_salary = (
    20000 +
    (years_experience * 30000) +
    (num_certifications * 10000) +
    np.random.normal(0, 10000, size=n_samples)  # noise
)
X = np.column_stack((years_experience, num_certifications))
y = monthly_salary
# 2. Train-test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)
# 3. Scale features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
# 4. Hyperparameter tuning with GridSearchCV
param_grid = {
    'n_neighbors': [3, 5, 7, 9],
    'weights': ['uniform', 'distance'],
    'p': [1, 2]  # Manhattan vs Euclidean
}
model = KNeighborsRegressor()
grid = GridSearchCV(model, param_grid, cv=5, scoring='r2')
grid.fit(X_train_scaled, y_train)
# 5. Evaluate
best_model = grid.best_estimator_
y_pred = best_model.predict(X_test_scaled)
print("✅ Best Parameters:", grid.best_params_)
print(f"📉 MSE: {mean_squared_error(y_test, y_pred):.2f}")
print(f"📈 R² Score: {r2_score(y_test, y_pred):.4f}")
# 6. Plot predicted vs actual salary
plt.figure(figsize=(8, 6))
plt.scatter(y_test, y_pred, alpha=0.6, c='navy')
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--')
plt.title("KNN Regression: Actual vs Predicted Salary")
plt.xlabel("Actual Salary (₹)")
plt.ylabel("Predicted Salary (₹)")
plt.grid(True)
plt.tight_layout()
plt.show()
'''Output -
✅ Best Parameters: {'n_neighbors': 5, 'p': 2, 'weights': 'distance'}
📉 MSE: 388274854.64
📈 R² Score: 0.9779
'''
```
