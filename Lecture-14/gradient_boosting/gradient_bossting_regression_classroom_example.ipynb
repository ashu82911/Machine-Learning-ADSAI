```python
import numpy as np
import pandas as pd
# Step 1: Create the dataset
data = pd.DataFrame({
   'Age': [22, 25, 28, 35, 45],
   'Experience': [1, 2, 4, 6, 10],
   'Income': [30, 35, 45, 60, 80]  # Target variable
})
# Step 2: Initial prediction - mean of target
F0 = data['Income'].mean()
print(f"Initial prediction (F0): {F0:.2f}")
# Step 3: Residuals for first iteration
data['r1'] = data['Income'] - F0
# Step 4: Fit first stump h1(x) based on Experience
split = 5
group1 = data[data['Experience'] <= split]
group2 = data[data['Experience'] > split]
r1_left = group1['r1'].mean()
r1_right = group2['r1'].mean()
def h1(exp):
   return r1_left if exp <= split else r1_right
# Step 5: Update F1
lr = 0.5
data['F1(x)'] = F0 + lr * data['Experience'].apply(h1)
# Step 6: Residuals for second iteration
data['r2'] = data['Income'] - data['F1(x)']
# Step 7: Fit second stump h2(x)
group1 = data[data['Experience'] <= split]
group2 = data[data['Experience'] > split]
r2_left = group1['r2'].mean()
r2_right = group2['r2'].mean()
def h2(exp):
   return r2_left if exp <= split else r2_right
# Step 8: Update F2
data['F2(x)'] = data['F1(x)'] + lr * data['Experience'].apply(h2)
# Step 9: Final residual
data['Final_Error'] = data['Income'] - data['F2(x)']
# Step 10: Predict new sample
new_sample = {'Age': 40, 'Experience': 8}
f1_pred = F0 + lr * h1(new_sample['Experience'])
f2_pred = f1_pred + lr * h2(new_sample['Experience'])
# Print results
print("\nFinal Data After Two Iterations:")
print(data[['Age', 'Experience', 'Income', 'F1(x)', 'F2(x)', 'Final_Error']])
print(f"\nPrediction for new sample {new_sample}:\nF1: {f1_pred:.2f}, F2: {f2_pred:.2f}")
```
