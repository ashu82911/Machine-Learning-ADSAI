```python
import tensorflow as tf
# Scalar (0D tensor)
scalar_tensor = tf.constant(5)
print(f"Scalar Tensor: {scalar_tensor}")
# Vector (1D tensor)
vector_tensor = tf.constant([1, 2, 3, 4])
print(f"Vector Tensor: {vector_tensor}")
# Matrix (2D tensor)
matrix_tensor = tf.constant([[1, 2], [3, 4]])
print(f"Matrix Tensors: {matrix_tensor}")
# 3D Tensor
tensor_3d = tf.constant([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
print(f"Tensors 3D: {tensor_3d}")
# Tensor of zeros (2D tensor)
zeros_tensor = tf.zeros([3, 3])
print(f"Zero Tensors: {zeros_tensor}")
# Tensor of ones (2D tensor)
ones_tensor = tf.ones([2, 2])
print(f"Ones Tensors: {ones_tensor}")
```
